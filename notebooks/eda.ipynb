# MovieLens 100K - Exploratory Data Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
%matplotlib inline
```

```python
movies = pd.read_csv('data/movies.csv')
ratings = pd.read_csv('data/ratings.csv')
tags = pd.read_csv('data/tags.csv')
links = pd.read_csv('data/links.csv')
```

```python
print('Movies shape:', movies.shape)
print('Ratings shape:', ratings.shape)
print('Tags shape:', tags.shape)
print('Links shape:', links.shape)
```

## Basic Statistics

```python
print(f'Number of unique users: {ratings["userId"].nunique()}')
print(f'Number of unique movies: {ratings["movieId"].nunique()}')
print(f'Total number of ratings: {len(ratings)}')
print(f'Average rating: {ratings["rating"].mean():.2f}')
```

## Rating Distribution

```python
plt.figure(figsize=(10, 6))
ratings['rating'].value_counts().sort_index().plot(kind='bar', color='steelblue')
plt.title('Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

## Most Rated Movies

```python
movie_ratings_count = ratings.groupby('movieId').size().sort_values(ascending=False).head(20)
movie_ratings_count = movie_ratings_count.reset_index(name='count')
movie_ratings_count = movie_ratings_count.merge(movies[['movieId', 'title']], on='movieId')

plt.figure(figsize=(12, 8))
plt.barh(movie_ratings_count['title'], movie_ratings_count['count'], color='coral')
plt.xlabel('Number of Ratings')
plt.title('Top 20 Most Rated Movies')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## Genre Distribution

```python
genres_series = movies['genres'].str.split('|').explode()
genre_counts = genres_series.value_counts().head(15)

plt.figure(figsize=(10, 6))
genre_counts.plot(kind='bar', color='seagreen')
plt.title('Genre Distribution')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Ratings per User

```python
user_ratings_count = ratings.groupby('userId').size()

plt.figure(figsize=(10, 6))
plt.hist(user_ratings_count, bins=50, color='mediumpurple', edgecolor='black')
plt.title('Ratings per User Distribution')
plt.xlabel('Number of Ratings')
plt.ylabel('Number of Users')
plt.show()

print(f'Average ratings per user: {user_ratings_count.mean():.1f}')
print(f'Min ratings per user: {user_ratings_count.min()}')
print(f'Max ratings per user: {user_ratings_count.max()}')
```

## Average Rating per Movie

```python
avg_ratings = ratings.groupby('movieId')['rating'].agg(['mean', 'count'])
avg_ratings = avg_ratings[avg_ratings['count'] >= 10]
avg_ratings = avg_ratings.merge(movies[['movieId', 'title']], on='movieId')

plt.figure(figsize=(10, 6))
plt.hist(avg_ratings['mean'], bins=30, color='goldenrod', edgecolor='black')
plt.title('Average Rating Distribution (Movies with 10+ ratings)')
plt.xlabel('Average Rating')
plt.ylabel('Count')
plt.show()
```
